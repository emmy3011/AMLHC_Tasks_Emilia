Load the Dataset

```python
import pandas as pd

# Load dataset
df_diabetes = pd.read_csv("diabetes.csv")

# Look at the first few rows to ensure everything looks correct
print(df_diabetes.head())
```

Fit a Generalized Linear Model (Logistic Regression)

```python
from sklearn.linear_model import LogisticRegression

# Separate features and target
X = df_diabetes.drop(columns=['class'])
y = df_diabetes['class']

# Initialize and fit the Logistic Regression model
model_logistic = LogisticRegression(max_iter=1000)
model_logistic.fit(X, y)

print("Logistic Regression Model successfully fitted!")
```

Evaluate the Model

```python
from sklearn.metrics import accuracy_score

# Extract coefficients and intercept
print("Intercept:", model_logistic.intercept_)
print("Coefficients (for each feature respectively):")
for col, coef in zip(X.columns, model_logistic.coef_[0]):
    print(f"  {col}: {coef:.4f}")

# Make predictions on the dataset
predictions = model_logistic.predict(X)

# Calculate accuracy score
acc = accuracy_score(y, predictions)
print(f"\nTraining Accuracy Score: {acc:.4f}")
```

Train and Evaluate using Cross-Validation & Split

```python
from sklearn.model_selection import train_test_split, cross_val_score

# 1. Fit on training data and evaluate on test data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=123)

# Retrain model on train set only
model_logistic.fit(X_train, y_train)
test_predictions = model_logistic.predict(X_test)
test_acc = accuracy_score(y_test, test_predictions)
print(f"Test Set Accuracy Score: {test_acc:.4f}")

# 2. 5-Fold Cross-Validation
# We recreate a fresh instance of the model for cross-validation
cv_scores = cross_val_score(LogisticRegression(max_iter=1000), X, y, cv=5)
print("\nCross-Validation Accuracy scores for each fold:", cv_scores)
print(f"Mean CV Accuracy Score: {cv_scores.mean():.4f}")
```
